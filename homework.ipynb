Part 2: Finding station names
----------------------------------------------------------------------------------------

```python
# Load station data
stations = pd.read_csv("data/indego-stations.csv")
stations.head()
```

```python
# Merge starting stations with rides dataframe
df = pd.merge(
    df, 
    stations[['Station_ID', 'Station_Name']],
    left_on='start_station',
    right_on='Station_ID',
    how='left'
)

df.head()
```

```python
# Rename starting station column & remove extra station ID column
df = df.rename(columns={'Station_Name': 'start_station_name'})
df = df.drop(columns=['Station_ID'])
df[['start_station', 'start_station_name']].head()
```

```python
# Merge ending stations with rides dataframe 
df = pd.merge(
    df,
    stations[['Station_ID', 'Station_Name']],
    left_on='end_station', 
    right_on='Station_ID',
    how='left'
)

df.head()
```

```python
# Rename ending station column & remove extra station ID column
df = df.rename(columns={'Station_Name': 'end_station_name'})
df = df.drop(columns=['Station_ID'])
df[['end_station', 'end_station_name']].head()
```

```python
print(df.columns)
print(f"Total rides: {len(df)}")
```

```python
# Extract year data from each ride's start time
df['Year'] = df['start_time'].dt.year
df[['start_time', 'Year', 'start_station_name']].head()
```

```python
# Yearly rides start at each station
yearly_start_rides = df.groupby(['start_station_name', 'Year'])['trip_id'].count()
yearly_start_rides.head(10)
```

```python
# Average yearly rides for each starting location
average_start_rides = yearly_start_rides.groupby('start_station_name').mean()
average_start_rides.head(10)
```

```python
# Standard deviation for yearly rides for each starting location
std_start_rides = yearly_start_rides.groupby('start_station_name').std()
std_start_rides.head(10)
```

```python
# Combine average and std into one dataframe
start_station_stats = pd.concat([average_start_rides, std_start_rides], axis=1)
start_station_stats.columns = ['Average', 'Standard Deviation']
start_station_stats.head(10)
```

```python
# Sort starting stations by average yearly rides
top_ten_start = start_station_stats.sort_values('Average', ascending=False)
top_ten_start.head(10)
```

```python
# Yearly rides ending at each station
yearly_end_rides = df.groupby(['end_station_name', 'Year'])['trip_id'].count()
yearly_end_rides.head(10)
```

```python
# Average yearly rides for each ending station
average_end_rides = yearly_end_rides.groupby('end_station_name').mean()
average_end_rides.head(10)
```

```python
# Standard deviation of yearly rides for each ending station
std_end_rides = yearly_end_rides.groupby('end_station_name').std()
std_end_rides.head(10)
```

```python
# Combine average and std into one dataframe
end_station_stats = pd.concat([average_end_rides, std_end_rides], axis=1)
end_station_stats.columns = ['Average', 'Standard Deviation']
end_station_stats.head(10)
```

```python
# Sort ending stations by average yearly rides
top_ten_end = end_station_stats.sort_values('Average', ascending=False)
top_ten_end.head(10)
```

Part 3: Rides per day-of-the-week
----------------------------------------------------------------------------------------

```python
# Extract date from the start_time
df['start_date'] = df['start_time'].dt.normalize()
df[['start_time', 'start_date']].head()
```

```python
# Extract the day of the week 
df['Day_of_week'] = df['start_date'].dt.dayofweek
df[['start_date', 'Day_of_week']].head()
```

```python
# Count number of rides started on each date
daily_rides = df.groupby('start_date')['trip_id'].count()
daily_rides.head(10)
```

```python
# Convert daily ride counts into dataframe, rename ride count column, add day of the week 
daily_rides = daily_rides.reset_index()
daily_rides = daily_rides.rename(columns={'trip_id': 'Rides'})
daily_rides['Day_of_week'] = daily_rides['start_date'].dt.dayofweek
daily_rides.head(10)
```

```python
# Average daily rides for each day of the week 
average_daily_rides = daily_rides.groupby('Day_of_week')['Rides'].mean()
average_daily_rides
```

```python
# Std of daily rides for each day of the week 
std_daily_rides = daily_rides.groupby('Day_of_week')['Rides'].std()
std_daily_rides
```

```python
# Combine average and std into one dataframe
weekday_stats = pd.concat([average_daily_rides, std_daily_rides], axis=1)
weekday_stats.columns = ['Average', 'Standard Deviation']
weekday_stats
```

Part 4: Rides and precipitation

```python
# Load weather data
weather = pd.read_csv("data/weather.csv")
weather.head()
```

```python
# Convert weather date column to datetime
weather['DATE'] = pd.to_datetime(weather['DATE'])
print(weather['DATE'].dtype)
weather[['DATE', 'PRCP', 'SNOW']].head()
```

```python
# Merge weather data with the rides dataframe
df = pd.merge(
    df,
    weather,
    left_on='start_date',
    right_on='DATE',
    how='left'
)

df[['start_date', 'DATE', 'PRCP', 'SNOW']].head()
```

```python
# Merge weather data with daily ride counts
daily_rides = pd.merge(
    daily_rides,
    weather,
    left_on='start_date',
    right_on='DATE',
    how='left'
)

daily_rides[['start_date', 'Rides', 'PRCP', 'SNOW']].tail(10)
```

```python
# Find days with no rain or snow, get average daily rides on dry days
dry_days = daily_rides[(daily_rides['PRCP'] == 0) & (daily_rides['SNOW'] == 0)]
average_dry_rides = dry_days['Rides'].mean()
print("Average rides on dry days:", average_dry_rides)
```

```python
# Days with rain 
rainy_days = daily_rides[daily_rides['PRCP'] > 0]
average_rainy_rides = rainy_days['Rides'].mean()
print("Average rides on rainy days:", average_rainy_rides)
```

```python
# Days with snow 
snowy_days = daily_rides[daily_rides['SNOW'] > 0]
average_snowy_rides = snowy_days['Rides'].mean()
print("Average rides on snowy days:", average_snowy_rides)
```

Part 6: Time series of ride times
----------------------------------------------------------------------------------------

```python
# Get hour from each ride's start time
df['Hour'] = df['start_time'].dt.hour
df[['start_time', 'Hour']].head(10)
```

```python
# Column for the part of the day
df['part_of_day'] = 'Unclassified'

# Morning -- 5 AM to 12 PM
df.loc[(df['Hour'] >= 5) & (df['Hour'] < 12), 'part_of_day'] = 'morning'

# Afternoon -- 12 PM to 5 PM
df.loc[(df['Hour'] >= 12) & (df['Hour'] < 17), 'part_of_day'] = 'afternoon'

# Evening -- 5 PM to 9 PM
df.loc[(df['Hour'] >= 17) & (df['Hour'] < 21), 'part_of_day'] = 'evening'

# Night -- 9 PM to 4 AM
df.loc[(df['Hour'] >= 21) | (df['Hour'] < 4), 'part_of_day'] = 'night'

df[['start_time', 'Hour', 'part_of_day']].head(10)
```

```python
# Exclude rides outside the four time ranges
classified_rides = df[df['part_of_day'] != 'Unclassified']

# Get rides by part of day and year 
morning_rides = classified_rides[classified_rides['part_of_day'] == 'morning'].groupby('Year')['trip_id'].count()

afternoon_rides = classified_rides[classified_rides['part_of_day'] == 'afternoon'].groupby('Year')['trip_id'].count()

evening_rides = classified_rides[classified_rides['part_of_day'] == 'evening'].groupby('Year')['trip_id'].count()

night_rides = classified_rides[classified_rides['part_of_day'] == 'night'].groupby('Year')['trip_id'].count()

morning_rides
```

```python
# Plot yearly rides for each part of the day 
plt.figure(figsize=(10, 6))
plt.plot(morning_rides.index, morning_rides, marker='o', label='morning')
plt.plot(afternoon_rides.index, afternoon_rides, marker='o', label='afternoon')
plt.plot(evening_rides.index, evening_rides, marker='o', label='evening')
plt.plot(night_rides.index, night_rides, marker='o', label='night')

plt.xlabel('Year')
plt.ylabel('Number of Rides')
plt.title('Indego Bike Rides by Part of Day (2018–2025)')

plt.xticks(morning_rides.index)
plt.legend(title='Part of Day')

plt.savefig('indego_part_of_day.png', dpi=300, bbox_inches='tight')
plt.show()
```
